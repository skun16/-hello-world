```python
warehouse_url = 'https://gz.58.com/baiyun/cangkucf/'
```

```python
page = WebPage()
# page.get(warehouse_url)
```

```python
def get_street_name():
    street_urls = []
    page.get('https://gz.58.com/cangkucf/?PGTID=0d100000-0000-31e0-2732-4feb2e6aefeb&ClickID=7')
    district_url_page = page.ele('@class=area-sub-title').eles('@name=b_link')
    district_urls = [url.link for url in district_url_page]
    for district_url in district_urls:
        page.get(district_url)
        street_urls.extend([url.link for url in page.ele('@class=area-sub-content clearfix').eles('@href')])
    return street_urls
        
```

```python
street_urls = get_street_name()
```

```python
import pickle
with open('street_urls.pkl','wb') as f:
    pickle.dump(street_urls, f)
```
